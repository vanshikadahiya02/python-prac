```python
student = {"name": "meera", "age": 20, "course": "physics"}
print(student["name"])
student["age"] = 21
student["city"] = "pune"
print(student)
```

```python
for key in student:
    print(key, " ", student[key])
```

```python
for k, v in student.items():
    print(k, v)
```

```python
phone = {"ram": 98231, "shyam": 90012}
print(phone.get("ram"))
print(phone.get("geeta", "not found"))
```

```python
sentence = "hellow world"
freq = {}
for ch in sentence:
    if ch == " ":
        continue
    freq[ch] = freq.get(ch, 0) + 1
print(freq)
```

```python
word = "mississippi"
freq = {}
for ch in word:
    freq[ch] = freq.get(ch, 0) + 1
print(freq)
```

```python
d1 = {"a": 1, "b": 2}
d2 = {"c": 3, "d": 4}
merged = {}
for k, v in d1.items():
    merged[k] = v
for k, v in d2.items():
    merged[k] = v
print(merged)
```

```python
scores = {"maths": 78, "science": 91, "english": 85}
best = None
top = -1
for subj, score in scores.items():
    if score > top:
        top = score
        best = subj
print(best)
```

```python
students = {
    "meera": [78, 85, 92],
    "arjun": [66, 72, 80],
    "kavya": [90, 88, 95],
}
for name, marks in students.items():
    avg = sum(marks) / len(marks)
    print(name, round(avg, 2))
```
