```python
def greet():
    print("hello there")

greet()
```

```python
def add(a, b):
    return a + b

print(add(4, 9))
```

```python
def describe(name, age=18):
    print(name, "is", age, "years old")

describe("neha")
describe("arjun", 22)
```

```python
def total(*args):
    s = 0
    for x in args:
        s += x
    return s

print(total(1, 2, 3, 4))
```

```python
double = lambda x: x * 2
print(double(7))
```

```python
def fact(n):
    if n <= 1:
        return 1
    return n * fact(n - 1)

print(fact(5))
```

```python
def is_even(n):
    return n % 2 == 0

print(is_even(4))
print(is_even(7))
```

```python
def rect_area(length, width=5):
    return length * width

print(rect_area(4))
print(rect_area(4, 3))
```

```python
def is_palindrome(word):
    return word == word[::-1]

print(is_palindrome("madam"))
print(is_palindrome("hello"))
```

```python
def biggest(a, b, c):
    if a >= b and a >= c:
        return a
    if b >= a and b >= c:
        return b
    return c

print(biggest(3, 9, 6))
```

```python
def welcome(name):
    return "hello, " + name

print(welcome("riya"))
```
