```python
items = ["a", "b", "a", "c", "b", "a"]
seen = []
for it in items:
    if it not in seen:
        seen.append(it)
print(seen)
```

```python
nums = [5, 9, 3, 12, 7, 12]
big = second = float("-inf")
for n in nums:
    if n > big:
        second = big
        big = n
    elif n > second and n != big:
        second = n
print(second)
```

```python
vals = [4, -2, 9, -7, 0, 3]
for i in range(len(vals)):
    if vals[i] < 0:
        vals[i] = 0
print(vals)
```
