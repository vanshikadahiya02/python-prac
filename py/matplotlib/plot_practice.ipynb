```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("simple line")
plt.show()
```

```python
subjects = ["maths", "science", "english", "history"]
scores = [88, 76, 92, 65]

plt.bar(subjects, scores)
plt.title("scores by subject")
plt.show()
```

```python
import random

ages = [random.randint(18, 60) for _ in range(50)]
heights = [random.randint(150, 190) for _ in range(50)]

plt.scatter(ages, heights)
plt.xlabel("age")
plt.ylabel("height")
plt.show()
```

```python
import numpy as np

data = np.random.randn(1000)
plt.hist(data, bins=30)
plt.title("histogram")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot([1, 2, 3], [1, 4, 9])
axes[0].set_title("square")
axes[1].bar(["a", "b", "c"], [3, 7, 5])
axes[1].set_title("bars")
plt.show()
```

```python
x = list(range(0, 11))
y = [v * v for v in x]

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("x squared")
plt.show()
```

```python
months = ["jan", "feb", "mar", "apr", "may", "jun"]
sales = [1200, 1500, 1100, 1800, 1650, 2000]

plt.bar(months, sales)
plt.title("monthly sales")
plt.show()
```

```python
x = list(range(1, 11))
y1 = [v for v in x]
y2 = [v * v for v in x]

plt.plot(x, y1, label="linear")
plt.plot(x, y2, label="square")
plt.legend()
plt.show()
```
