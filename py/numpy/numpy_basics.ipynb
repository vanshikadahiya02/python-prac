```python
import numpy as np

arr = np.array([1, 2, 3, 4])
print(arr)
print(arr * 2)
```

```python
z = np.zeros((2, 3))
o = np.ones(4)
r = np.arange(0, 10, 2)
print(z)
print(o)
print(r)
```

```python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print(matrix.shape)
print(matrix[1, 2])
print(matrix[:, 1])
```

```python
rows = np.arange(12).reshape(3, 4)
print(rows)
rows = rows + 10
print(rows)
```

```python
data = np.array([23, 45, 67, 12, 98, 34])
print(data.mean())
print(data.sum())
print(data.max())
print(data.std())
```

```python
m = np.zeros((5, 5))
for i in range(5):
    m[i, i] = 1
print(m)
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(np.dot(a, b))
```

```python
arr = np.array([5, 8, 12, 7, 20, 3])
print(arr[arr % 2 == 0])
```

```python
arr = np.arange(1, 13)
print(arr.reshape(3, 4))
```
