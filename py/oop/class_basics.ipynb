```python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

    def bark(self):
        print(self.name, "says woof!")

d = Dog("tommy", "labrador")
d.bark()
```

```python
class Circle:
    pi = 3.14159

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return Circle.pi * self.radius ** 2

c = Circle(5)
print(c.area())
```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(self.name, "makes a sound")

class Cat(Animal):
    def speak(self):
        print(self.name, "meows")

a = Animal("generic")
a.speak()
kitty = Cat("kitty")
kitty.speak()
```

```python
class BankAccount:
    def __init__(self, holder, balance=0):
        self.holder = holder
        self.balance = balance

    def deposit(self, amt):
        self.balance += amt

    def withdraw(self, amt):
        if amt <= self.balance:
            self.balance -= amt
            return True
        return False

acc = BankAccount("rohan", 1000)
acc.deposit(500)
print(acc.balance)
acc.withdraw(300)
print(acc.balance)
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

p = Point(3, 4)
print(p)
```

```python
print("1. rectangle class with area and perimeter methods")
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

r = Rectangle(4, 5)
print(r.area())
print(r.perimeter())
```

```python
print("2. student class with a marks list and an average method")
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def average(self):
        return sum(self.marks) / len(self.marks)

s = Student("neha", [78, 85, 92])
print(s.average())
```

```python
print("3. counter class that starts at 0 and can increment and reset")
```

```python
class Counter:
    def __init__(self):
        self.value = 0

    def increment(self):
        self.value += 1

    def reset(self):
        self.value = 0

c = Counter()
c.increment()
c.increment()
print(c.value)
c.reset()
print(c.value)
```

```python
print("4. shape base class with a circle subclass that overrides area")
```

```python
class Shape:
    def area(self):
        return 0

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

c = Circle(3)
print(c.area())
```

```python
print("5. book class with title, author, pages and a nice print")
```

```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __str__(self):
        return self.title + " by " + self.author + " (" + str(self.pages) + " pages)"

b = Book("the alchemist", "paulo coelho", 208)
print(b)
```
