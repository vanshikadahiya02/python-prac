```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "total_bill": [16.99, 10.34, 21.01, 23.68, 24.59, 25.29, 8.77, 26.88, 15.04, 14.78],
    "tip": [1.01, 1.66, 3.50, 3.31, 3.61, 4.71, 2.00, 3.12, 1.96, 3.23],
    "sex": ["male", "female", "male", "male", "female", "male", "female", "male", "female", "female"],
    "day": ["sun", "sun", "sun", "fri", "sat", "sat", "thu", "thu", "fri", "sat"],
    "smoker": ["no", "yes", "no", "no", "yes", "yes", "no", "yes", "no", "yes"],
}
df = pd.DataFrame(data)
print(df.head())
```

```python
sns.scatterplot(data=df, x="total_bill", y="tip")
plt.title("bill vs tip")
plt.show()
```

```python
sns.histplot(df["total_bill"], bins=8)
plt.show()
```

```python
sns.boxplot(data=df, x="day", y="tip")
plt.show()
```

```python
sns.countplot(data=df, x="day")
plt.show()
```

```python
df["size"] = [2, 3, 2, 4, 3, 5, 2, 3, 4, 2]
sns.scatterplot(data=df, x="size", y="tip")
plt.show()
```

```python
sns.histplot(df["tip"], bins=6)
plt.show()
```

```python
sns.boxplot(data=df, x="sex", y="total_bill")
plt.show()
```

```python
sns.countplot(data=df, x="smoker")
plt.show()
```

```python
sns.lmplot(data=df, x="total_bill", y="tip")
plt.show()
```
