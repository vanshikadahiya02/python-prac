```python
import streamlit as st

st.title("my first app")
st.write("hello from streamlit")
```

```python
name = st.text_input("your name")
if name:
    st.write("hi", name)
```

```python
number = st.slider("pick a number", 1, 100, 25)
st.write("double is", number * 2)
```

```python
import pandas as pd

df = pd.DataFrame({"a": [1, 2, 3], "b": [4, 5, 6]})
st.dataframe(df)
```

```python
option = st.selectbox("choose a fruit", ["apple", "mango", "banana"])
st.write("you chose", option)
```

```python
n = st.slider("number", 1, 20, 5)
st.write("square is", n * n)
```

```python
nm = st.text_input("enter your name")
if nm:
    st.write("hello,", nm)
```

```python
show = st.checkbox("show secret message")
if show:
    st.write("you found it!")
```

```python
city = st.selectbox("pick a city", ["delhi", "mumbai", "chennai"])
st.write("you picked", city)
```

```python
a = st.number_input("first number", value=0)
b = st.number_input("second number", value=0)
st.write("sum is", a + b)
```
